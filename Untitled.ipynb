```python
import rasterio 
import numpy as np
```

```python
file_path = "/Users/ehsan/my_landsat_image/rgb.tif"
```

```python
with rasterio.open(file_path) as sourceData:
    print(f"Success! Opened: {file_path}")
    image = sourceData.read()
    
```

```python
bands,rows,columns = image.shape[0],image.shape[1],image.shape[2],
```

```python
number_pixels = rows * columns
```

```python
reshaped_image = image.reshape( 3, number_pixels).T
centered_reshape = reshaped_image - np.mean(reshaped_image , axis = 0)
```

```python
c = np.cov(centered_reshape , rowvar=False)
```

```python
print (c)
```

```python
from sklearn.decomposition import PCA
 
pca = PCA()
pca.fit(reshaped_image)

# Eigenvalues = explained variance
print("Eigenvalues:", pca.explained_variance_)

# Eigenvectors = components
print("Eigenvectors (first PC):", pca.components_[0])
```

```python
print("Eigenvectors (first PC):", pca.components_[1])
```

```python
print("Eigenvectors (first PC):", pca.components_[2])
```
